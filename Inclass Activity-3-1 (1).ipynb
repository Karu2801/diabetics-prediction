# 1. Loading the Dataset
## We'll use Seaborn to load the Titanic dataset:

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

# Load Titanic dataset
titanic = sns.load_dataset('titanic')
```

# 2. Initial Exploration
## Let's start with the basic EDA: checking missing values, data types, and getting a glimpse of the data.

```python

# Initial Exploration
# Checking for missing values
print(titanic.isnull().sum())

```

```python
# Data information
print(titanic.info())
```

```python
# Basic statistics
print(titanic.describe())
```

# 3. Exploratory Data Analysis (EDA)

## 3.1. Bar Plot of Survival
## A quick overview of how many people survived or didn't.

## In this section you have to generate all kinds of plots, you can choose any variable that is reasonable.

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(x='survived', data=titanic)
plt.title('Survival Distribution')
plt.show()
```

## Understanding the age distribution of the passengers.

```python
sns.histplot(titanic['age'].dropna(), kde=True, bins=30)
plt.title('Age Distribution')
plt.show()
```

## 3.3. Survival Rate by Class
## Visualizing how survival rates vary with passenger class.

```python
sns.barplot(x='pclass', y='survived', data=titanic)
plt.title('Survival Rate by Passenger Class')
plt.show()
```

## 3.4. Heatmap of Correlations
## A heatmap to show correlations between numerical variables.

```python
numeric_columns = titanic.select_dtypes(include=['float64', 'int64'])

# Plotting the heatmap of correlations
sns.heatmap(numeric_columns.corr(), annot=True, cmap='coolwarm')
plt.title('Heatmap of Correlations')
plt.show()
```

```python
sns.boxplot(x='pclass', y='age', data=titanic)
plt.title('Age Distribution by Passenger Class')
plt.show()
```

```python
sns.boxplot(x='survived', y='fare', data=titanic)
plt.title('Fare Distribution by Survival')
plt.show()
```

```python
sns.barplot(x='sex', y='survived', data=titanic)
plt.title('Survival Rate by Gender')
plt.ylabel('Survival Rate')
plt.show()
```

```python
sns.scatterplot(x='age', y='survived', data=titanic, alpha=0.5)
plt.title('Survival Rate by Age')
plt.ylabel('Survived (0 = No, 1 = Yes)')
plt.show()
```

```python
sns.countplot(x='pclass', hue='survived', data=titanic)
plt.title('Count of Passengers by Class and Survival')
plt.xlabel('Passenger Class')
plt.ylabel('Count')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()
```

```python
survival_class_gender = titanic.groupby(['pclass', 'sex'])['survived'].mean().unstack()
sns.heatmap(survival_class_gender, annot=True, cmap='YlGnBu')
plt.title('Survival Rate by Class and Gender')
plt.xlabel('Gender')
plt.ylabel('Passenger Class')
plt.show()
```

```python
sns.boxplot(x='survived', y='fare', data=titanic)
plt.title('Fare Distribution by Survival')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Fare')
plt.show()
```

```python
sns.pairplot(titanic, hue='survived', vars=['age', 'fare', 'pclass'])
plt.suptitle('Pairplot of Numerical Features Colored by Survival', y=1.02)
plt.show()
```

```python
sns.violinplot(x='pclass', y='age', data=titanic)
plt.title('Age Distribution by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Age')
plt.show()
```

```python
# Data Preprocessing for Classification

# Ensure 'age' is numeric before filling missing values
titanic['age'] = pd.to_numeric(titanic['age'], errors='coerce')
titanic['age'] = titanic['age'].fillna(titanic['age'].median())

# Dropping rows where 'embarked' is missing
titanic.dropna(subset=['embarked'], inplace=True)

if 'deck' in titanic.columns:
    # Add 'Unknown' as a category to 'deck'
    titanic['deck'] = titanic['deck'].cat.add_categories('Unknown')
    # Fill missing values in 'deck' with 'Unknown'
    titanic['deck'] = titanic['deck'].fillna('Unknown')

# Check for any remaining missing values
print(titanic.isnull().sum())

# Encoding Categorical Variables
titanic_encoded = pd.get_dummies(titanic, columns=['sex', 'embarked', 'class', 'deck', 'who', 'alone'], drop_first=True)

# Dropping unnecessary columns
titanic_encoded.drop(['embark_town', 'alive', 'adult_male'], axis=1, inplace=True)

# Checking the processed data
print(titanic_encoded.head())
```
