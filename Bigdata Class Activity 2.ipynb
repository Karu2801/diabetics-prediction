**1) Creating a DataFrame**

Explanation: Imported pandas and created a DataFrame which inculdes car details :Car_ID, Make, Model, Colour.

```python
import pandas as pd
# Define data as a dictionary
data = {
       'Car_ID' :[1,2,3,4,5],
       'Make'   :['Toyota','Honda','Ford','Tesla','BMW'],
       'Model'  :['Camry','Civic','Mustang','Model 3','X5'],
       'Colour' :['White','Black','Red','Blue','Gray'],
    'Price'  :[24425.02,21250.08,43000.02,39990.01,60600.04]
}
#Create DataFrame
df = pd.DataFrame(data)
print(df)
```

**2. Choosing Data and Filtering**


Selecting specific coloumns and processing the rows based on particular area of data.

```python
# Select 'Make' and 'Model' columns
selected_columns = df[['Make', 'Model']]
print(selected_columns)
print("\n")
print("\n")
# Filter rows where 'Price($)' is greater than 25000
filtered_data = df[df['Price'] > 25000]
print(filtered_data)
```

**3) Clustering**

Explanation :

Computing total sales for each Model by multiplying price by,and then aggregate these sales by Model.

```python
# Calculate total sales
df['Total_sales'] = df['Car_ID'] * df['Price']
print(df)
print("\n")
print("\n")
aggregate_sales = df.groupby('Model').Total_sales.sum()
print(aggregate_sales)
```

**4) Classify Data**

Explanation: 

Classifying the Data Frame based on the Total_sales,in both ascending order and desending order.


```python
# Sort by 'Total_sales' in ascending order
sorted_ascending = df.sort_values(by='Total_sales')
print(sorted_ascending)
print("\n")
print("\n")
# Sort by 'Total_sales' in descending order
sorted_descending = df.sort_values(by='Total_sales', ascending=False)
print(sorted_descending)
```

**5) Handling Missing Values**

Explanation:

Handling the data where it is incomplete , such as missing Price by replacing the missing values with the average price for each model.


```python
# Simulate missing value
df.loc[5, 'Price'] = pd.NA
# Fill missing values with the average price per product
average_prices = df.groupby('Model')['Price'].transform('mean')
print(average_prices)
print("\n")
print("\n")
df['Price'] = df['Price'].fillna(average_prices)
print(df)
```

**6)Integrating DataFrames**

Explanation:

Merging Make data with another dataset,such as customer rating to improve the analysis



```python
# New DataFrame for rating
rating_data = {
'Make': ['Toyota', 'Ford', 'Tesla','BMW'],
'Rating': ['5', '4', '4.5','3']
}
rating_df = pd.DataFrame(rating_data)
print(rating_df)
print("\n")
print("\n")
# Merge Make data with rating
merged_df = pd.merge(df, rating_df, on='Make', how='right')
print(merged_df)
```

**7) Complex Aggregations**

Explanation:

Conduct more comprehensive aggregations by organizing the data based on multiple categories or dimensions.



```python
# Group and calculate Price
grouped_Price = df.groupby(['Make', 'Colour']).agg(
Total_Price=('Total_sales', 'sum'),
Average_Price=('Total_sales', 'mean')
)
print(grouped_Price)
```

**8) Data Transformation**

Explanation:

Implement data transformations, such as applying discounts according to the Price.

```python
# Apply discount if Price > 3
def apply_discount (row):
    if row['Price'] > 4:
        return row ['Price'] * 0.9 #10% discount
    return row['Price']

df['Discounted_Price'] = df.apply(apply_discount, axis=1)
print (df)
```

**9) Generating Conditional Columns**

Explanation:

Classify Price volume based on the Car_ID to easily identify high and low Price volumes.



```python
# Categorize Price volume
df['Price_Volume'] = df['Car_ID'].apply(lambda x: 'High' if x > 3 else 'Low')
print(df)
```
