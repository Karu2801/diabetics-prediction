```python
import pandas as kt
import matplotlib.pyplot as plt
import numpy as nm
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
file_path = "mtcars.csv" 
mtcars = kt.read_csv(r"C:\Users\karun\Downloads\mtcars.csv")
```

```python
mtcars = mtcars[['mpg', 'wt']]
```

```python
plt.scatter(mtcars['wt'], mtcars['mpg'], color='Green', label="Data centers")
plt.xlabel("Vehicle Weight (wt)")
plt.ylabel("Gas Mileage (mpg)")
plt.title("Scatter Plot of Vehicle Weight vs Gas Mileage")
plt.legend()
plt.show()
```

```python
X = mtcars[['wt']]
y = mtcars['mpg']
linear_model = LinearRegression()
linear_model.fit(X, y)
```

```python
intercept = linear_model.intercept_
coefficient = linear_model.coef_[0]
```

```python
y_pred = linear_model.predict(X)
```

```python
plt.scatter(mtcars['wt'], mtcars['mpg'], color='Yellow', label="Data centers")
plt.plot(mtcars['wt'], y_pred, color='Black', label="Linear Regression Line")
plt.xlabel("Vehicle Weight (wt)")
plt.ylabel("Gas Mileage (mpg)")
plt.title("Linear Regression: Vehicle Weight vs Gas Mileage")
plt.legend()
plt.show()
```

```python
mse_linear = mean_squared_error(y, y_pred)
print(f"Linear Model: Intercept = {intercept}, Coefficient = {coefficient}, MSE = {mse_linear}")
```

```python
X_squared = kt.DataFrame({'wt': mtcars['wt'], 'wt_squared': mtcars['wt']**2})
quad_model = LinearRegression()
quad_model.fit(X_squared, y)
```

```python
intercept_quad = quad_model.intercept_
coefficients_quad = quad_model.coef_
y_quad_pred = quad_model.predict(X_squared)
```

```python
plt.scatter(mtcars['wt'], mtcars['mpg'], color='brown', label="Data centers")
sorted_indices = nm.argsort(mtcars['wt'])
plt.plot(mtcars['wt'].iloc[sorted_indices], y_quad_pred[sorted_indices], color='blue', label="Quadratic Regression Curve")
plt.xlabel("Vehicle Weight (wt)")
plt.ylabel("Gas Mileage (mpg)")
plt.title("Quadratic Regression: Vehicle Weight vs Gas Mileage")
plt.legend()
plt.show()
```

```python
mse_quadratic = mean_squared_error(y, y_quad_pred)
print(f"Quadratic Model: Intercept = {intercept_quad}, Coefficients = {coefficients_quad}, MSE = {mse_quadratic}")
```

```python
Linear Regression Model:
Intercept: Value of estimated mpg when wt is 0.
Coefficient: Value of change in mpg if wt is raised by one unit.
Mean Squared Error (MSE): Signifies the fit of the model (better smaller).
```

```python
Intercept: Value changed considering second-order term.
Coefficients: Two, one for wt and one for wt^2.
Mean Squared Error (MSE): How well the quadratic model fits.
```
