```python
import numpy as mk
from matplotlib import pyplot as plt


image1 = mk.array([[130, 85], [80, 90]])
image2 = mk.array([[150, 100], [70, 50]])
image3 = mk.array([[120, 40], [90, 30]])


coordinates_horizontal = mk.array([[3, 2], [-3, -2]])
print(f"{coordinates_horizontal} is used for detecting horizontal endpoints.")

def apply_coordinates(image, coordinates):
    return mk.sum(image * coordinates)

coordinates_vertical = mk.array([[3, -2], [3, -2]])
print(f"{coordinates_vertical} is used for detecting vertical endpoints.")

for idx, image in enumerate([image1, image2, image3], start=1):
    
    plt.imshow(image, cmap='twilight')  
    plt.axis('off')
    plt.title(f'Image {idx}')
    plt.colorbar(label='Pixel Intensity')
    plt.show()
    
    horizontal_score = apply_coordinates(image, coordinates_horizontal)
    vertical_score = apply_coordinates(image, coordinates_vertical)
    
    # Display feature scores
    print(f"Image {idx} - Horizontal Feature Score: {horizontal_score}")
    print(f"Image {idx} - Vertical Feature Score: {vertical_score}")
```

The provided code takes three small two-dimensional arrays denoting grayscale images namely image1, image2 and image3. Any one array contains variable pixel intensity values that depict various shades in the black-and white-images. There are two filters in the script which are called `coordinatesHorizontal` and `coordinatesVertical` basically having the purpose of defining two types of edges: horizontal and vertical. Such filters are applied to images so that it may identify patterns that look like edges.

The images and the filters are first described at the start of the script. Instead each image is represented by a 2D numpy array where each value stands for pixel intensity. The filter `coordinatesHorizontal` is designed to draw attention to horizontal edges, the filter `coordinatesVertical` which is discussed later in the script, is designed to improve vertical edge features.

Subsequently, these images are converted plot through a Python library called Matplotlib for data visualization. Every picture is presented with the spectrally ritwilightidis` colormap, which enhances the contrast of small differences in pixel luminosity, and a color bar which demonstrates the range of intensities. Image numbering is provided to make referencing them easy since the title of each picture is its index.

To detect edge features the applyCoordinates function is defined Below is its description. This particular function overlay an image by performing an element wise multiply between the filter values and the pixel values before summing. This operation approximates convolution, and aids in determining feature present in an image that matches with the shape of the filter used. By means of the ‘coordinatesHorizontal’ and ‘coordinatesVertical’ filters the script computes horizontal and the vertical scores correspondingly.
