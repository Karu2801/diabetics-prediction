```
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# Load dataset
file_path = "mtcars.csv"
mtcars = pd.read_csv(file_path)

# Selecting relevant columns
mtcars = mtcars[['mpg', 'wt']]

# Scatter plot of the dataset
plt.scatter(mtcars['wt'], mtcars['mpg'], color='blue', label="Data points")
plt.xlabel("Vehicle Weight (wt)")
plt.ylabel("Gas Mileage (mpg)")
plt.title("Scatter Plot of Vehicle Weight vs Gas Mileage")
plt.legend()
plt.show()
```

```
# Linear Regression Model
X = mtcars[['wt']]
y = mtcars['mpg']
linear_model = LinearRegression()
linear_model.fit(X, y)

# Get model parameters
intercept = linear_model.intercept_
coefficient = linear_model.coef_[0]

# Predictions
y_pred = linear_model.predict(X)

# Plot the regression line
plt.scatter(mtcars['wt'], mtcars['mpg'], color='blue', label="Data points")
plt.plot(mtcars['wt'], y_pred, color='red', label="Linear Regression Line")
plt.xlabel("Vehicle Weight (wt)")
plt.ylabel("Gas Mileage (mpg)")
plt.title("Linear Regression: Vehicle Weight vs Gas Mileage")
plt.legend()
plt.show()

# Calculate Mean Squared Error (MSE)
mse_linear = mean_squared_error(y, y_pred)

(intercept, coefficient, mse_linear)
```

```
# Adding a squared term for weight
X_squared = pd.DataFrame({'wt': mtcars['wt'], 'wt_squared': mtcars['wt']**2})
quad_model = LinearRegression()
quad_model.fit(X_squared, y)

# Get new model parameters
intercept_quad = quad_model.intercept_
coefficients_quad = quad_model.coef_

# Predictions for quadratic model
y_quad_pred = quad_model.predict(X_squared)

# Plot the quadratic regression curve
plt.scatter(mtcars['wt'], mtcars['mpg'], color='blue', label="Data points")
sorted_indices = np.argsort(mtcars['wt'])
plt.plot(mtcars['wt'].iloc[sorted_indices], y_quad_pred[sorted_indices], color='green', label="Quadratic Regression Curve")
plt.xlabel("Vehicle Weight (wt)")
plt.ylabel("Gas Mileage (mpg)")
plt.title("Quadratic Regression: Vehicle Weight vs Gas Mileage")
plt.legend()
plt.show()

# Calculate Mean Squared Error (MSE) for quadratic model
mse_quadratic = mean_squared_error(y, y_quad_pred)

(intercept_quad, coefficients_quad, mse_quadratic)
```
