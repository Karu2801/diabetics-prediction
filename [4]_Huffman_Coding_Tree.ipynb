```python
import heapq
```

```python
class Node:
    def __init__(self, char=None, freq=None):
        self.char = char
        self.freq = freq
        self.left = None
        self.right = None

    def __lt__(self, other):
        if self.freq == other.freq:
            return ord(self.char) < ord(other.char)
        return self.freq < other.freq
```

```python
def build_huffman_tree(char_freqs):
    heap = [Node(char, freq) for char, freq in char_freqs.items()]
    heapq.heapify(heap)

    while len(heap) > 1:
        node1 = heapq.heappop(heap)
        node2 = heapq.heappop(heap)
        merged_node = Node(freq=node1.freq + node2.freq)
        merged_node.left = node1
        merged_node.right = node2
        heapq.heappush(heap, merged_node)

    return heapq.heappop(heap)
```

```python
def generate_huffman_codes(node, current_code="", codes={}):
    if node is None:
        return

    if node.char is not None:
        codes[node.char] = current_code
        return

    generate_huffman_codes(node.left, current_code + "0", codes)
    generate_huffman_codes(node.right, current_code + "1", codes)

    return codes
```

```python
def print_tree(node, level=0, prefix="Root: "):
    if node is not None:
        print("  " * level + prefix + f"({node.freq})", end="")
        if node.char:
            print(f" - {node.char}")
        else:
            print()
        print_tree(node.left, level + 1, "L: ")
        print_tree(node.right, level + 1, "R: ")
```

```python
char_freqs = {}
while True:
    input_str = input("Enter character and frequency (e.g., A 6), or press Enter to finish: ")
    if not input_str:
        break
    char, freq = input_str.split()
    char = char.upper()
    char_freqs[char] = char_freqs.get(char, 0) + int(freq)

root = build_huffman_tree(char_freqs)
huffman_codes = generate_huffman_codes(root)

print("\nHuffman Codes:")
for char, code in huffman_codes.items():
    print(f"{char}: {code}")

print("\nHuffman Tree:")
print_tree(root)
```
