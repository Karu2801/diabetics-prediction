# Implementation 3: Data Preprocessing

### Data Preprocessing



Handle missing values appropriately if any (e.g., filling missing data or removing rows/columns). 

```python
yulu['temp'] = yulu['temp'].fillna(yulu['temp'].median())
```

```python
yulu.head()
```

### Convert Categorical Variables

Convert any categorical columns into numerical values using techniques like one-hot encoding.

```python
yulu_encoded = kv.get_dummies(yulu, columns=['casual', 'count', 'datetime', 'atemp'], drop_first=False)
```

```python
yulu_encoded.head()
```

###     Drop Unnecessary Columns



Remove columns that do not contribute to the model or are irrelevant.

```python
yulu_encoded.drop(['casual_4', 'atemp_37.88', 'atemp_40.91'], axis=1, inplace=True)
```

# Implementation 4: Train-Test Split & Model Training

### Split the dataset into a training set and a testing set. 

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
```

```python
k = yulu_encoded.drop('registered', axis=1)
r = yulu_encoded['registered']
```

### Train 2 Machine Learning Models



Select any two machine learning algorithms from Logistic Regression, Random Forest, KNN, SVM, Neural Networks, DBSCAN, or even fine-tune a pre-trained model.


```python
k_train, k_test, r_train, r_test = train_test_split(k, r, test_size=0.3, random_state=42)
```

```python
knn = KNeighborsClassifier()
knn.fit(k_train, r_train)
```

```python
rf = RandomForestClassifier(n_estimators=120)
rf.fit(k_train, r_train)
```

# Implementation 5: Cross-Validation and Model Evaluation

### Evaluate Your Models



Calculate precision, recall, F1-score, and accuracy for both models.


```python
from sklearn.metrics import classification_report, precision_score, recall_score, f1_score
```

### Perform Cross-Validation

Use cross-validation to check the stability of your models across different subsets of the data.

```python
from sklearn.model_selection import cross_val_score
```

```python
r_pred_rf = rf.predict(k_test)
print("Random Forest Classification Report:\n", classification_report(r_test, r_pred_rf))
```

```python
r_pred_knn = knn.predict(k_test)
print("KNN Classification Report:\n", classification_report(r_test, r_pred_knn))
```
